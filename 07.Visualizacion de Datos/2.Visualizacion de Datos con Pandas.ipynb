```python
import pandas as pd
```

```python
# leer el archivo csv
df_population_raw = pd.read_csv('population_total.csv')
```

```python
# mostrar el dataframe
df_population_raw
```

# Hacer una Tabla Pivote

.pivot(): Devuelve un dataframe reformado que esta organizado por cierto index/valores de columnas ("tabla pivote sin funciones agregadas) 

```python
# eliminar valores nulos
df_population_raw.dropna(inplace=True)
```

```python
# hacer tabla pivote
df_pivot = df_population_raw.pivot(index='year', columns='country',
                                        values='population')
```

```python
# seleccionar algunos paises
df_pivot = df_pivot[['United States', 'India', 'China', 
                          'Indonesia', 'Brazil']]
```

```python
# mostrar tabla pivote
df_pivot
```

# Lineplot (Grafico de linea)

```python
df_pivot.plot(kind='line',
              xlabel='Years', ylabel='Population',
              title='Population (1955-2020)',
              figsize=(8,4))
```

# Barplot (Grafico de caja)

## Barplot Separados

```python
# seleccionando solo un año (2020)
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
# mostrando un dataframe
df_pivot_2020
```

```python
# trasponer un dataframe (cambiando filas con columnas y viceversa)
df_pivot_2020 = df_pivot_2020.T
```

```python
# hacer el barplot
df_pivot_2020.plot(kind='bar', color='orange',
                   xlabel='Years', ylabel='Population',
                   title='Population in 2020')
```

## Barplot agrupado por “n” variables

```python
# seleccionar unos años
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020])]
```

```python
# haciendo barplot agrupados
df_pivot_sample.plot(kind='bar',
                     xlabel='Years', ylabel='Population',
                     figsize=(8,4))
```

# Piechart

```python
# cambiando el nombre de columna
df_pivot_2020.rename(columns={2020:'2020'}, inplace=True)
```

```python
# haciendo el piechart
df_pivot_2020.plot(kind='pie',
                   y='2020',
                   title='Population in 2020 (%)')
```

# Boxplot

## Boxplot

```python
df_pivot['United States'].plot(kind='box', color='green', 
                               ylabel='Population')
```

## Multiples Boxplot

```python
df_pivot.plot(kind='box', xlabel='Countries',
              ylabel='Population')
```

# Histograma

```python
df_pivot[['United States', 'Indonesia']].plot(kind='hist',
                                              xlabel='Population')
```

# ScatterPlot (Grafico de dispersion)

```python
# seleccionar una muestra de la data
df_sample = df_population_raw[df_population_raw['country'].isin(['United States', 'India', 'China', 'Indonesia', 'Brazil'])]
```

```python
# haciendo el plot
df_sample.plot(kind='scatter',
               x='year',
               y='population',
               figsize=(8,5),
               color='green',
               s=80)
```
